```python
-- 1. Tworzenie tabeli dla placówek sieci Medica+
CREATE TABLE Kliniki (
    id_kliniki INT PRIMARY KEY AUTO_INCREMENT,
    nazwa VARCHAR(100) NOT NULL,
    adres VARCHAR(255) NOT NULL
);

-- 2. Tworzenie tabeli Pacjenci (atrybuty: imię, nazwisko, PESEL, data urodzenia)
CREATE TABLE Pacjenci (
    id_pacjenta INT PRIMARY KEY AUTO_INCREMENT,
    imie VARCHAR(50) NOT NULL,
    nazwisko VARCHAR(50) NOT NULL,
    pesel CHAR(11) UNIQUE NOT NULL,
    data_urodzenia DATE NOT NULL
);

-- 3. Tworzenie tabeli Lekarze (atrybuty: specjalizacja, numer prawa wykonywania zawodu)
CREATE TABLE Lekarze (
    id_lekarza INT PRIMARY KEY AUTO_INCREMENT,
    imie VARCHAR(50) NOT NULL,
    nazwisko VARCHAR(50) NOT NULL,
    specjalizacja VARCHAR(100) NOT NULL,
    nr_prawa_zawodu CHAR(7) UNIQUE NOT NULL
);

-- 4. Tworzenie tabeli Wizyty (relacje: pacjent może mieć wiele wizyt, wizyta przypisana do lekarza i kliniki)
CREATE TABLE Wizyty (
    id_wizyty INT PRIMARY KEY AUTO_INCREMENT,
    id_pacjenta INT,
    id_lekarza INT,
    id_kliniki INT,
    data_wizyty DATETIME NOT NULL,
    status VARCHAR(20) DEFAULT 'Zarejestrowana',
    FOREIGN KEY (id_pacjenta) REFERENCES Pacjenci(id_pacjenta),
    FOREIGN KEY (id_lekarza) REFERENCES Lekarze(id_lekarza),
    FOREIGN KEY (id_kliniki) REFERENCES Kliniki(id_kliniki)
);
```
